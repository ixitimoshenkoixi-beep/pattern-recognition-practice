## 3. Завантаження набору даних

```python
iris = load_iris(as_frame=True)
print(type(iris))
print(iris.keys())
```

```python
print("Назви ознак:")
for feature_name in iris.feature_names:
  print("-", feature_name)
print("\nНазви класів:")
for target_name in iris.target_names:
  print("-", target_name)
```

```python
df = iris.frame.copy()
display(df.head())
```

```python
target_mapping = {
 0: "setosa",
 1: "versicolor",
 2: "virginica"
}
print(target_mapping)
```

## 6. Візуалізація даних

```python
feature_columns = [
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm"
    ]

data[feature_columns].hist(
    bins=15,
    figsize=(10, 8)
    )

plt.suptitle("Розподіли числових ознак набору Iris")
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

1. Яка ознака має найменші значення?

petal_width_cm (значення починаються приблизно від $0.1$ см).

2. Яка ознака має найбільший діапазон?

petal_length_cm (значення розкидані від
1.0 до
6.9 см, тобто діапазон становить близько
5.9 см). Для порівняння, у sepal_length_cm діапазон становить від
4.3 до
7.9 см (близько
3.6 см).

3. Чи всі розподіли мають однакову форму?

Ні, не всі. Ознака sepal_width_cm має розподіл, близький до нормального (дзвоноподібного). Натомість petal_length_cm та petal_width_cm є явно бімодальними (мають два окремі «піки» або кластери), а sepal_length_cm має більш рівномірний мультимодальний розподіл.

 4. Для яких ознак можна помітити кілька груп значень?

petal_length_cm та petal_width_cm. На їхніх гістограмах чітко видно розрив та поділ на дві окремі групи (перша група з малими значеннями, потім розрив/значний пропад, і друга група з більшими значеннями).


```python
class_counts = data["species"].value_counts().sort_index()

class_counts.plot(
    kind="bar",
    figsize=(8, 5)
    )

plt.title("Кількість об’єктів кожного класу")
plt.xlabel("Вид ірису")
plt.ylabel("Кількість об’єктів")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

Набір даних є ідеально збалансованим. Як видно з графіка, кожен із трьох класів («setosa», «versicolor» та «virginica») містить рівно по 50 об'єктів

```python
for species_name, group in data.groupby("species"):
  plt.scatter(
      group["petal_length_cm"],
      group["petal_width_cm"],
      label=species_name,
      alpha=0.8
    )

  plt.title("Залежність довжини та ширини пелюстки")
  plt.xlabel("Довжина пелюстки, см")
  plt.ylabel("Ширина пелюстки, см")
  plt.legend()
  plt.grid(alpha=0.3)
  plt.show()

```

1. Чи утворюють об’єкти різних класів окремі групи?

Так. Об'єкти класів утворюють розрізнювані кластери в просторі ознак довжини та ширини пелюстки.

2. Який клас відокремлюється найкраще?

setosa. Значення цього класу розташовані в діапазонах $1.0\text{–}1.9$ см для довжини та $0.1\text{–}0.6$ см для ширини пелюстки, що повністю відокремлює його від інших двох класів без жодного перекриття.

3. Які два класи частково перекриваються?

versicolor та virginica. Їхні значення перекриваються на межі довжини пелюстки близько $4.5\text{–}5.1$ см та ширини пелюстки близько $1.4\text{–}1.8$ см.   

4. Чи можна використати ці ознаки для розпізнавання виду квітки?

Так, безумовно. Довжина та ширина пелюстки є дуже інформативними ознаками. Вони дозволяють зі $100\%$ точністю відокремити клас setosa, а також з високою точністю розділити класи versicolor та virginica (за допомогою лінійного або нелінійного класифікатора).
